**Q1: Simulate 30 rolls with =RANDBETWEEN(1,6). What is the probability of rolling a 3 exactly 5 times? (Hint: Use BINOM.DIST)**
- Given:
  - n = 30
  - k = 5
  - p(3) = 1/6

- Solution:
  - By using excel formula: 
    - probability of rolling a 3 exactly 5 times = "**=BINOM.DIST(5, 30, 1/6, FALSE)**"
    - probability of rolling a 3 exactly 5 times = **0.192108**

**Q3: A dataset has a mean of 50 and a standard deviation of 5. What percentage of values lie between 45 and 55 if the data follows a normal distribution?**
- Between 45 and 55 is ±1 standard deviation from the mean (50 ± 5).
- In a normal distribution, about 68% of values lie within ±1 standard deviation.
- **Answer: ≈ 68%**

**Q4: What is the concept of standardization (z-score), and why is it important in data analysis? Explain the formula and how standardization transforms a dataset.**
- Standardization (Z-score) is the process of converting a value into the number of standard deviations it is from the mean.
- Formula: 𝑍=(𝑋−𝜇)/𝜎

- Where:
  - X = original value
  - μ = mean
  - σ = standard deviation

- **Why it’s important:**
  - Allows comparison between different datasets
  - Converts data to a common scale
  - Helps identify outliers
  - Makes data suitable for many statistical methods

- **Effect of Standardization:**
  - New mean = 0
  - New standard deviation = 1
  - Values show how far they are from the mean in standard deviation units

**Q5: What is Kurtosis and their type?**
- Kurtosis measures the peakedness and tail heaviness of a distribution compared to a normal distribution.
- Types of Kurtosis:
  - Mesokurtic
    - Normal level of peakedness
    - Example: Normal distribution
    - Kurtosis ≈ 3
  
  - Leptokurtic
    - More peaked
    - Heavy tails (more outliers)
    - Kurtosis > 3
  
  - Platykurtic
    - Flatter peak
    - Light tails (fewer extreme values)
    - Kurtosis < 3

**Q6: Explain why the uniform distribution is a good model for the outcome of rolling a fair die.**
- A uniform distribution is appropriate for rolling a fair die because:
  - A fair die has 6 possible outcomes (1–6)
  - Each outcome has an equal probability
  - Probability of each number = 1/6

**Q8: How does log transformation help in stabilizing variance and making data more normally distributed?**
- Log transformation reduces skewness and stabilizes variance by compressing large values more than small values.

- How it helps:
  - Reduces right skewness
  - Makes spread more constant (stabilizes variance)
  - Minimizes impact of extreme values
  - Makes data closer to normal distribution
